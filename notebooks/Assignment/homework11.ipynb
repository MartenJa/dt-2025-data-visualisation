Lastly, I created something pretty similar to what we did in our Thursday's class. I linked my previos text and then an image mask from our class, because I liked the shape of it. After that I wrote a 'WordCloud' code so it would render out an image full of words in a shape of a circle. Lastly I also added some stopwords (I tried using an official list of estonian stopwords, but it didn't seem to work for some reason, so I typed in some of my own).

```python
import urllib.request
from PIL import Image #This is so it could render out images
import numpy as np #This is for the code could use the image mask to shape my wordcloud

text = urllib.request.urlopen("https://raw.githubusercontent.com/MartenJa/dt-2025-data-visualisation/refs/heads/codespace-jubilant-orbit-gr97595599x39g76/notebooks/Assignment/text").read().decode("utf-8")

img = urllib.request.urlretrieve("https://drive.google.com/uc?export=download&id=1axqWQpZOo_baYKv3XkI7ohtBtK956mX2", "image1.png")

from wordcloud import WordCloud

stopwords_list = urllib.request.urlopen("https://datadoi.ee/bitstream/handle/33/78/estonian-stopwords.txt?sequence=1&isAllowed=y").read().decode("utf-8").split("\n")


WordCloud(stopwords=["jah", "ning", "et", "sest", "kuid", "kui", "ja", "oli", "kus", "kes", "mis", "teiselt", "ühelt"], mask=np.array(Image.open("image1.png")), background_color="pink").generate_from_text(text).to_image()
```
