```python
text = urllib.request.urlopen("https://raw.githubusercontent.com/MartenJa/dt-2025-data-visualisation/refs/heads/codespace-jubilant-orbit-gr97595599x39g76/notebooks/class_work/test").read().decode("utf-8")

img = urllib.request.urlretrieve("https://drive.google.com/uc?export=download&id=1axqWQpZOo_baYKv3XkI7ohtBtK956mX2", "image1.png")

stopwords_list = urllib.request.urlopen("https://gist.githubusercontent.com/sebleier/554280/raw/7e0e4a1ce04c2bb7bd41089c9821dbcf6d0c786c/NLTK's%2520list%2520of%2520english%2520stopwords").read().decode("utf-8").split("\n")
# The file that we are downloadingcontains 1 word per 1 line. When python takes it, it'll become 1 long string (this\nand\nthat). So split("\n") will cut the text every time it sees a new line 

WordCloud(mask=np.array(Image.open("image1.png")), background_color="pink").generate_from_text(text).to_image()
```

```python
def color_function(word, font_size, position, orientation, random = None, **additional_args): #def color_function(...): This creates a function named color_function
    if len(word) < 4: #If the word has fewer than 4 letters, then the color  is green. If or more that 4, then blue.
        return "green"
    return "blue"

WordCloud().generate_from_text(text).recolor(color_func=color_function).to_image()
```
