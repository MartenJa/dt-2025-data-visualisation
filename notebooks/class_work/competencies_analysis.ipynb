```python
import pandas as pd

df = pd.read_csv("Competencies.csv")
df
```

```python
import matplotlib.pyplot as plt

avg_hours_studied = df.groupby("Student")["Hours_Studied"].mean()
avg_hours_studied.plot(kind="bar")

#.mean tähendab, et me kalkuleerime avg

plt.title("Hours studied per student")
plt.ylabel("Hours Studied")
plt.show()
```

```python
import matplotlib.pyplot as plt

avg_grades = df.groupby("Student")["Grade"].mean().sort_values(ascending=False)
avg_grades.plot(kind="bar")

#sort_values(ascending=False) tähendab, et see paneb numbrid kahanevasse järjekorda

plt.title("Who performs best?")
plt.ylabel("Grade")
plt.xlabel("Name")
plt.show()
```

```python
avg_grades = df.groupby("Student")["Grade"].mean().sort_values(ascending=False)

plt.figure(figsize=(8,8))  # to chenge the size of the chart

#plot the pie chart
plt.pie(avg_grades, labels=avg_grades.index, autopct='%1.1f%%', startangle=140, colors=plt.cm.Paired.colors)

#colors=plt.cm.Paired.colors = A coor map to give different colors to different slices
#startangle=140 = Rotate the chart to make more appealing (optional)
#autopct='%1.1f%%' = This format the percentages to one decimal place (withot it there's no percentages)
#, labels=avg_grades.index = adds a label to each section

plt.show
```

```python
plt.scatter(df["Hours_Studied"], df["Grade"],s=df["Hours_Studied"]*30, alpha=0.6)
plt.xlabel("Hours Studied")
plt.ylabel("Grade")
plt.title("Does more studying = better grades?")
plt.show()
```

```python
import matplotlib.pyplot as plt
#Which course has the highest average student performance?
avg_course = df.groupby("Course")["Grade"].mean()
print("Average Grade per course: \n", avg_course, "\n")

avg_course.plot(kind="bar", figsize=(8,5), color="green")
plt.ylabel("Average grade per course")
plt.title("Bar chart for average performance per course")
plt.show()

#Which program has the highest average student performance?
avg_program = df.groupby("Program")["Grade"].mean()
print("Average Grade per program: \n", avg_program, "\n")

avg_program.plot(kind="bar", figsize=(6,4), color="gold")
plt.ylabel("Average grade per program")
plt.title("Bar chart for average performance per program")
plt.show()
```

```python
hours_min = df["Hours_Studied"].min()
hours_q1 = df["Hours_Studied"].quantile(0.25)
hours_q2 = df["Hours_Studied"].quantile(0.50)
hours_q3 = df["Hours_Studied"].quantile(0.75)
hours_max = df["Hours_Studied"].max()

#Years of practice

grade_min = df["Grade"].min()
grade_q1 = df["Grade"].quantile(0.25)
grade_q2 = df["Grade"].quantile(0.50)
grade_q3 = df["Grade"].quantile(0.75)
grade_max = df["Grade"].max()

print("Hours_Studied - Min:", hours_min, "Q1", hours_q1, "Q3", hours_q3, "Max", hours_max)
print("Grade - Min:", grade_min, "Q1", grade_q1, "Q3", grade_q3, "Max", grade_max)

plt.figure(figsize=(8,5))
df[["Hours_Studied", "Grade"]].boxplot()
plt.title("Box plot to find how consictent are study paterns and performance")
plt.show
```
