```python
import pandas as pd

df = pd.read_csv("favourite_movies_v2.csv")
df
```

```python
import matplotlib.pyplot as plt

duration = df.groupby("Film")["Duration (min)"].mean().sort_values(ascending=False)
duration.plot(kind="bar", x="Film", y="Duration (min)", legend=False)
plt.title("Duration per movie")
plt.xlabel("Movies")
plt.ylabel("Duration (min)")
plt.show()
```

With this column chart I found out, that the longest movie is Hamlet.

```python
plt.scatter(df["Ilmumisaasta"], df["Duration (min)"], s=df["Ilmumisaasta"]*0.5, alpha=0.6)
plt.xlabel("Duration")
plt.ylabel("Ilmumisaasta")
plt.title("Were longer movies (>200 min) made before of after 2000s?")
plt.show()
```

In this Bubble chart I realised, that the longest movies from the list were made before 2000s.

```python
duration = df.groupby("Film")["Reiting"].mean().sort_values(ascending=False)

plt.figure(figsize=(8,8)) 
plt.title("Pie chart for every movie's rating")
plt.pie(duration, labels=duration.index, autopct='%1.1f%%', startangle=140, colors=plt.cm.Paired.colors)
```

This Pie chart showed me, that although many movies have almost similar ratings, but the most has Parasite.

```python
#calculate quartiles and extremes

rate_min = df["Reiting"].min()
rate_q1 = df["Reiting"].quantile(0.25)
rate_q2 = df["Reiting"].quantile(0.50)
rate_q3 = df["Reiting"].quantile(0.75)
rate_max = df["Reiting"].max()

#Years of practice

duration_min = df["Duration (min)"].min()
duration_q1 = df["Duration (min)"].quantile(0.25)
duration_q2 = df["Duration (min)"].quantile(0.50)
duration_q3 = df["Duration (min)"].quantile(0.75)
duration_max = df["Duration (min)"].max()

print("Reiting - Min:", rate_min, "Q1:", rate_q1, "Q2", rate_q2, "Q3:", rate_q3, "Max:", rate_max)

print("Duration (min) - Min:", duration_min, "Q1", duration_q1, "Q2", duration_q2, "Q3", duration_q3, "Max", duration_max)

#Drw box plot
plt.figure(figsize=(8,5))
df[["Reiting", "Duration (min)"]].boxplot()
plt.title("Box plot for rating and duration")
plt.show
```

In this chart, it's hard to really read any data from Rating, because the chart is either too small, or the numbers are too little.
